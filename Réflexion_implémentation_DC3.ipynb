```python
T = [97, 98, 99, 97, 98, 99, 97, 99, 97, 98, 0, 0, 0]
P_0 = []


for i in range(0, len(T)-2) :
    if i%3 == 0 : 
        P_0.append(i)

def P (P):
    return [x+1 for x in P]

P_1 = P(P_0)    
P_2 = P(P_1)
P_12 = P_1  + P_2 

print(P_0, P_1, P_2, P_12)
```

```python
def DC3(T) :
    T = [ord(char) for char in T] #convert string to num 
    n = len(T)
    T += [0, 0, 0] #add centinel numbers
    P_0 = [i for i in range(n+1) if i%3 == 0] #multiple de 3
    P_1 = [i+1 for i in P_0 if i<n] #P_0 + 1
    P_2 = [i+1 for i in P_1 if i<n] #P_1 + 1
    P_12 = P_1 + P_2 #concaténation P_12
    R_12 = [tuple(T[i:i + 3]) for i in P_12] #tuples à partir index P_12
    R_12s = sorted(R_12) #remplacer par radix sort plus tard
    
    return T, P_0, P_1, P_2, P_12, R_12, R_12s

print(DC3("abcabcacab"))
```

```python
def radix_sort(R): 
    for chiffre in (2, 1, 0): 
        max_value = max(triplet[chiffre] for triplet, _ in R)
    return max_value

R = [(98, 99, 97), (98, 99, 97), (99, 97, 98), (0, 0, 0), (99, 97, 98), (99, 97, 99), (97, 98, 0)]

print(radix_sort(R))
```
